### Aim
To implement the Dyna-Q reinforcement learning algorithm and study the effect of different numbers of planning steps ($n = 0, 5, 50$) on maze navigation performance, comparing the number of steps required to reach the goal.

```
import numpy as np
import matplotlib.pyplot as plt
from dyna_maze.environment import ROWS, COLS, START, GOAL, OBSTACLES
from dyna_maze.agent import DynaQ
```

### 1. Maze Environment Visualization
Visualizing the 6x9 GridWorld with obstacle barriers.

```
grid = np.zeros((ROWS, COLS))
for (r, c) in OBSTACLES: grid[r, c] = -1
grid[GOAL[0], GOAL[1]] = 2
grid[START[0], START[1]] = 1

plt.figure(figsize=(8, 4.5))
plt.imshow(grid, cmap='coolwarm')
plt.title('Dyna Maze Layout (6x9 GridWorld)', fontweight='bold')
plt.show()
```

### 2. Running Dyna-Q Across Planning Steps (n = 0, 5, 50)

```
n_planning_list = [0, 5, 50]
n_episodes = 50
n_runs = 10
results = {}

for n in n_planning_list:
    run_steps = np.zeros((n_runs, n_episodes))
    for run in range(n_runs):
        agent = DynaQ(n_planning=n)
        for ep in range(n_episodes):
            run_steps[run, ep] = agent.run_episode()
    results[n] = run_steps
    print(f"n={n}: Mean steps at Ep 2={run_steps[:, 1].mean():.1f}, Ep 50={run_steps[:, -1].mean():.1f}")
```

### 3. Log-Scale Performance Curve

```
episodes = np.arange(1, n_episodes + 1)
plt.figure(figsize=(9, 5))
for n in n_planning_list:
    plt.plot(episodes, results[n].mean(axis=0), lw=2, label=f'Dyna-Q (n={n})')
plt.yscale('log')
plt.title('Steps per Episode vs Planning Steps (Dyna-Q)', fontweight='bold')
plt.xlabel('Episode')
plt.ylabel('Steps to Goal (log scale)')
plt.legend()
plt.grid(True, which='both', alpha=0.3)
plt.tight_layout()
plt.show()
```

### 4. Conclusion & Findings
Dyna-Q integrates model-free Q-learning with model-based planning. When $n=50$, the agent discovers optimal trajectories in drastically fewer episodes compared to $n=0$, confirming the computational efficiency of planning in model-based RL.
